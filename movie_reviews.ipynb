STAGE 2:PREPROCESSING


```python
import re
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

nltk.download('punkt')
nltk.download('stopwords')
```

```python
print(df['review'][0])
```

```python
def clean_text(text):
    # Remove HTML tags
    text = re.sub(r'<.*?>', ' ', text)

    # Replace non-alphabet characters with space
    text = re.sub(r'[^a-zA-Z]', ' ', text)

    # Convert to lowercase
    text = text.lower()

    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()

    return text
```

```python
df['cleaned_review'] = df['review'].apply(clean_text)
```

```python
nltk.download('punkt_tab')
```

```python
df['tokens'] = df['cleaned_review'].apply(word_tokenize)
```

```python
stop_words = set(stopwords.words('english'))

# Keep important negation words
negations = {'not', 'no', 'nor'}
stop_words = stop_words - negations
```

```python
def remove_stopwords(tokens):
    return [word for word in tokens if word not in stop_words]
```

```python
df['filtered_tokens'] = df['tokens'].apply(remove_stopwords)
```

```python
print("ORIGINAL:\n", df['review'][0])
print("\nCLEANED:\n", df['cleaned_review'][0])
print("\nTOKENS:\n", df['tokens'][0])
print("\nAFTER STOPWORD REMOVAL:\n", df['filtered_tokens'][0])
```

STAGE 3: MORPHOLOGICAL ANALYSIS


```python
from nltk.stem import PorterStemmer, WordNetLemmatizer
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()
```

```python
def apply_stemming(tokens):
    return [stemmer.stem(word) for word in tokens]

df['stemmed_tokens'] = df['filtered_tokens'].apply(apply_stemming)
```

```python
def apply_lemmatization(tokens):
    return [lemmatizer.lemmatize(word) for word in tokens]

df['lemmatized_tokens'] = df['filtered_tokens'].apply(apply_lemmatization)
```

```python
print("ORIGINAL TOKENS:\n", df['filtered_tokens'][0])
print("\nSTEMMED TOKENS:\n", df['stemmed_tokens'][0])
print("\nLEMMATIZED TOKENS:\n", df['lemmatized_tokens'][0])
```

STAGE 7: SEMANTIC ANALYSIS

```python
import nltk
nltk.download('wordnet')
```

```python
from nltk.corpus import wordnet

sample_words = df['filtered_tokens'][0][:5]
print("Sample Words:", sample_words)

for word in sample_words:
    synsets = wordnet.synsets(word)
    if synsets:
        print(word, ":", synsets[0].definition())
```

```python
sentiment_words = ['good', 'bad', 'amazing', 'boring']

for word in sentiment_words:
    synsets = wordnet.synsets(word)
    if synsets:
        print("\nWord:", word)
        print("Meaning:", synsets[0].definition())
```

```python
word = "good"

synonyms = set()

for syn in wordnet.synsets(word):
    for lemma in syn.lemmas():
        synonyms.add(lemma.name())

print("Synonyms of 'good':", synonyms)
```
